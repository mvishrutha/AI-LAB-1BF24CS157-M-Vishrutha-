```python

import re

# Step 1: Eliminate implications
def eliminate_implications(expr):
    expr = re.sub(r'(\w+)\s*->\s*(\w+)', r'(~\1 | \2)', expr)
    return expr


# Step 2: Move NOT inward (De Morgan's law)
def move_not_inward(expr):
    expr = expr.replace('~((', '~(')

    while '~(~' in expr:
        expr = expr.replace('~(~', '')

    expr = expr.replace('~(P & Q)', '(~P | ~Q)')
    expr = expr.replace('~(P | Q)', '(~P & ~Q)')

    return expr


# Step 3: Skolemization
def skolemization(expr):
    expr = re.sub(r'exists\s+(\w+)', r'', expr)
    return expr


# Step 4: Remove universal quantifiers
def remove_universal(expr):
    expr = re.sub(r'forall\s+\w+\s*:\s*', '', expr)
    return expr


# Step 5: Distribute OR over AND
def distribute_or(expr):

    expr = expr.replace('(A | (B & C))',
                        '((A | B) & (A | C))')

    return expr


# Main CNF conversion
def convert_to_cnf(expression):

    print("\nOriginal FOL Expression:")
    print(expression)

    expression = eliminate_implications(expression)
    print("\nAfter eliminating implications:")
    print(expression)

    expression = move_not_inward(expression)
    print("\nAfter moving NOT inward:")
    print(expression)

    expression = skolemization(expression)
    print("\nAfter Skolemization:")
    print(expression)

    expression = remove_universal(expression)
    print("\nAfter removing universal quantifiers:")
    print(expression)

    expression = distribute_or(expression)
    print("\nFinal CNF Expression:")
    print(expression)


# Main program
print("FIRST ORDER LOGIC TO CNF CONVERSION")

fol = input("Enter FOL Expression: ")

convert_to_cnf(fol)

```
