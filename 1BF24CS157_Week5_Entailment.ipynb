```python

from itertools import product

# Define propositions
symbols = ['P', 'Q', 'R']

# Knowledge Base (KB)
# Example:
# P -> Q
# Q -> R
# P is true

def KB(model):
    P = model['P']
    Q = model['Q']
    R = model['R']

    return ((not P or Q) and
            (not Q or R) and
            P)

# Query
def query(model):
    return model['R']

# Check entailment
def entails():
    for values in product([True, False], repeat=len(symbols)):

        model = dict(zip(symbols, values))

        if KB(model):
            print("Model:", model)

            if not query(model):
                print("Query is NOT entailed by KB")
                return

    print("Query IS entailed by KB")


entails()
```
